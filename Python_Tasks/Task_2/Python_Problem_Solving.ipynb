## Q1. Find the Shortest Word in a Sentence

**Observe:** Input: 'Python makes coding easy'. Expected output: 'easy'.

**Understand:** Find the word with the fewest characters in the sentence.

**Break:** Split the sentence into words.

Assume the first word is the shortest.

Compare each word with the current shortest word.

Update the shortest word when a smaller one is found.

Print the result.

**Logic:** Use a loop to check each word's length. If a word is shorter than the current shortest word, replace it.

```python
sentence = "Python makes coding easy"
words = sentence.split()
shortest = words[0]
for word in words:
    if len(word) < len(shortest):
        shortest = word
print(shortest)

# Extra test
sentence2 = "Learning Python is fun"
words2 = sentence2.split()
shortest2 = words2[0]
for word in words2:
    if len(word) < len(shortest2):
        shortest2 = word
print(shortest2)
```

## Q4. Find the First Longest Word When Ties Occur

**Observe:** Input: 'Python makes coding useful'. Expected output: 'Python'.

**Understand:** We need to find the longest word in the sentence. If multiple words have the same maximum length, we must return the word that appears first.

**Break:** Store the sentence.

Split the sentence into words.

Assume the first word is the longest.

Compare the length of each word with the current longest word.

Update only when a word is strictly longer.

Print the result.

**Logic:** Use a for loop to check each word. If the current word has more characters than the longest word found so far, update the longest word. If their lengths are equal, do not update it, so the first word is preserved.

```python
sentence = "Python makes coding useful"
words= sentence.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)

# Additional test
sentence2 = "apple mango banana"
words2 = sentence2.split()
longest2 = words2[0]
for word in words2:
    if len(word) > len(longest2):
        longest2 = word
print(longest2)
```

## Q8. Find the Most Frequent Character

**Observe:** Input: 'banana'. Expected output: 'a'.

**Understand:** Find the value asked for in the question.

**Break:** Split/check the input, then compare or count values.

**Logic:** I ignore spaces and count each character. I choose the character with the highest count; max keeps the first tied character based on the original order of the keys.

**Test:** The first run uses the example input. A second test is included in the code below.

```python
text = "banana"
chars = [char for char in text if char != " "]
counts = {}
for char in chars:
    counts[char] = counts.get(char, 0) + 1
most_frequent = max(counts, key=counts.get)
print(most_frequent)

# Extra test
text2 = "aabb"
counts2 = {}
for char in text2.replace(" ", ""):
    counts2[char] = counts2.get(char, 0) + 1
print(max(counts2, key=counts2.get))
```

## Q10. Find All Characters Appearing More Than Once

**Observe:** Input: 'programming'. Expected output: '['r', 'g', 'm']'.

**Understand:** Find the value asked for in the question.

**Break:** Split/check the input, then compare or count values.

**Logic:** I go through the characters in order. If a character has already appeared and is not in the repeated list, I add it once.

**Test:** The first run uses the example input. A second test is included in the code below.

```python
text = "programming"
seen = set()
repeated = []
for char in text:
    if char == " ":
        continue
    if char in seen and char not in repeated:
        repeated.append(char)
    seen.add(char)
print(repeated)

# Extra test
text2 = "hello"
seen2 = set()
repeated2 = []
for char in text2:
    if char in seen2 and char not in repeated2:
        repeated2.append(char)
    seen2.add(char)
print(repeated2)
```

## Q15. Dictionary of Numbers and Their Even/Odd Status

**Observe:** Input: "[1, 2, 3, 4, 5]". Expected output: "{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}".

**Understand:** Find the value asked for in the question.

**Logic:** A number is even when its remainder after division by 2 is zero. Otherwise it is odd.

**Test:** The first run uses the example input. A second test is included in the code below.

```python
numbers = [1, 2, 3, 4, 5]
result = {}
for number in numbers:
    if number % 2 == 0:
        result[number] = "Even"
    else:
        result[number] = "Odd"
print(result)

# Extra test
numbers2 = [0, -3]
print({number: ("Even" if number % 2 == 0 else "Odd") for number in numbers2})
```

## Q20. Find the Longest Word That Appears Only Once

**Observe:** Input: 'Python makes coding useful Python'. Expected output: 'coding'.

**Understand:** Find the value asked for in the question.

**Break:** Split/check the input, then compare or count values.

**Logic:** I count lowercase words, keep words that appear once, and choose the longest. If lengths tie, the first one is kept.

**Test:** The first run uses the example input. A second test is included in the code below.

```python
sentence = "Python makes coding useful Python"
words = sentence.split()
lower_words = [word.lower() for word in words]
counts = {}
for word in lower_words:
    counts[word] = counts.get(word, 0) + 1
unique_words = [word for word in words if counts[word.lower()] == 1]
longest = max(unique_words, key=len)
print(longest)

# Extra test
sentence2 = "red orange blue"
words2 = sentence2.split()
counts2 = {}
for word in words2:
    counts2[word.lower()] = counts2.get(word.lower(), 0) + 1
unique2 = [word for word in words2 if counts2[word.lower()] == 1]
print(max(unique2, key=len))
```
